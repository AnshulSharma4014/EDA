```python
from ipywidgets import interact

def plotit(Country = 'India'):
    data_columns = ['2020', '2021', '2022', '2023', '2024', '2025']
    df2 = df.copy()
    df2 = df2[df2['Country'].str.lower() == Country.lower()]

    if len(df2) > 0:
        plt.figure(figsize=(9, 5))
        
        # Sum the values across the selected rows (in case a country appears multiple times)
        # and extract the time-series values.
        plot_values = df2[data_columns].values.flatten() #df2[data_columns].values.flatten().sum(axis=0)

        plt.plot(
            data_columns, 
            plot_values, 
            label=Country,
            #marker='o', # Add a marker for clarity
            #linestyle='--'
        )
        
        plt.title(f'GDP Trend for {Country}')
        plt.xlabel('Year')
        plt.ylabel('GDP (Trillions USD)') # You should specify the unit if you know it (e.g., 'Value (Millions USD)')
        plt.legend()
        plt.grid(axis='y', linestyle=':')
        plt.show()
    else:
        print(f"No data to show for: {Country}")

# Use interact to automatically create and display the Dropdown widget
# It recognizes the list of options and builds the appropriate widget.
interact(
    plotit, 
    Country=list(df['Country'].unique())
)
```

# Now Prepare the data for the ML training. 

1. The data is away from the mean as seen on the first graph
2. We need to use median for the null values (as the data is skewed).

```python
df.isnull().sum()
```

```python
df['2021'].median()
```

```python
df['2021'].fillna(df['2021'].median(), inplace=True)
df['2022'].fillna(df['2022'].median(), inplace=True)
df['2023'].fillna(df['2023'].median(), inplace=True)
df['2024'].fillna(df['2024'].median(), inplace=True)
df['2025'].fillna(df['2025'].median(), inplace=True)
```

```python
X = df.drop('Country', axis = 1)
y = df['Country']
```

```python
X.head(2)
```

```python
y.head(2)
```

# Map each country to integer

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
print(y)
```

```python
# Test train split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.33, random_state=42)
```

## Scale the training data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print("\n--- Scaling Summary ---")
print(f"X_train mean (pre-scaling): {X.values.mean():.2f}")
print(f"X_train_scaled mean: {X_train.mean():.2f}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.metrics import classification_report
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

Y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, Y_pred)
rmse = mse**0.5

print("\n--- Model Evaluation ---")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
```

```python
# Convert regression predictions to integer class labels
y_pred_labels = np.round(Y_pred).astype(int)
print(classification_report(y_test, y_pred_labels))
```
